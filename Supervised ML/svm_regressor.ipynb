```python
import pandas as pd 
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
from sklearn.metrics import r2_score
```

```python
df = datasets.load_diabetes(as_frame = True).frame 
```

```python
df.head()
df.shape
```

```python
X = df.drop("target", axis = 1)
y = df["target"]
```

```python
y  # ---> NOT SCALED
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.3, random_state = 42  
)
```

```python
#y_test and y_train can't be scaled directly. It expects 2D array.
# 'values.reshape(-1,1)' --> converts it to 2D aaray
# '.ravel()'--> converts 2D array to 1D array
y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled = y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
y_train.values.reshape(-1,1) # for my refernce only
```

```python
y_train.values.reshape(-1,1).ravel() # for my refernce only
```

```python
# SVR Model
model = SVR() #default -> 'rbf' for kernel
model.fit(X_train, y_train_scaled)
```

```python
# Predicting for both train and test data
y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)
```

```python
# Printing r2 score for both train and test data
print("train r2 score : ", r2_score( y_train_scaled , y_train_pred_scaled ))
print("test r2 score : ", r2_score( y_test_scaled , y_test_pred_scaled )) # OVERFITTING
```

```python
# Linear Kernel
model = SVR(kernel="linear")

model.fit(X_train, y_train_scaled)

y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)

print("train r2 score : ", r2_score( y_train_scaled , y_train_pred_scaled ))
print("test r2 score : ", r2_score( y_test_scaled , y_test_pred_scaled )) # NOT overfitting
```

```python
# Polynomial Kernel
model = SVR(kernel="poly")

model.fit(X_train, y_train_scaled)

y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)

print("train r2 score : ", r2_score( y_train_scaled , y_train_pred_scaled ))
print("test r2 score : ", r2_score( y_test_scaled , y_test_pred_scaled )) # OVERFITTING
```

```python
# Sigmoid Kernel
model = SVR(kernel="sigmoid")

model.fit(X_train, y_train_scaled)

y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)

print("train r2 score : ", r2_score( y_train_scaled , y_train_pred_scaled ))
print("test r2 score : ", r2_score( y_test_scaled , y_test_pred_scaled )) #VERY BAD CASE , DON'T CONSIDER
```
