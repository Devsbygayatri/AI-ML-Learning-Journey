# 1) HANDLE MISSING VALUES

```python
# Separating categorical and numerical cols
categorical_cols = df.select_dtypes(include = "object").columns
numerical_cols = df.select_dtypes(include = "number").columns # "number"->["float64" , "int64"]
```

```python
categorical_cols.size + numerical_cols.size # checking if we have separated all cols
```

```python
# SimpleImputer -> Replace missing values using a descriptive statistic 
from sklearn.impute import SimpleImputer

# for numerical
num_imp = SimpleImputer(strategy = "mean")
num_imp.fit_transform(df[numerical_cols]) # scaling
#df[numerical_cols] = num_imp.fit_transform(df[numerical_cols])  # assigning it in all num cols of df
```

```python
df[numerical_cols] = num_imp.fit_transform(df[numerical_cols]) 
df.head() # all NaN vals are replaced with mean for numerical cols
```

```python
# for categorical
cat_imp = SimpleImputer(strategy = "most_frequent")
df[categorical_cols] = cat_imp.fit_transform(df[categorical_cols]) 
df.head()# all NaN vals are replaced with mode /"most_frequent"  for categorical cols
```
