## Q 3. Consider the marks of 5 students in 4 subjects.  
[[85,90,88,92], 
[70,65,72,68], 
[95,91,93,96], 
[60,75,70,80], 
[88,84,86,90]] 
Find: 
1. Average marks of each student 
2. Average marks of each subject 
3. Highest and lowest marks 
4. Student having the highest total marks

```python
import numpy as np

# Marks of 5 students in 4 subjects
marks = np.array([[85, 90, 88, 92],
                  [70, 65, 72, 68],
                  [95, 91, 93, 96],
                  [60, 75, 70, 80],
                  [88, 84, 86, 90]])

# 1. Average marks of each student (mean across columns, axis=1)
avg_student = np.mean(marks, axis=1)

# 2. Average marks of each subject (mean across rows, axis=0)
avg_subject = np.mean(marks, axis=0)

# 3. Highest and lowest marks
highest_mark = np.max(marks)
lowest_mark = np.min(marks)

# 4. Student having the highest total marks
total_marks = np.sum(marks, axis=1)
best_student = np.argmax(total_marks) + 1  # Adding 1 for human-readable 1-based indexing

print("1. Average marks of each student:", avg_student)
print("2. Average marks of each subject:", avg_subject)
print("3. Highest mark:", highest_mark, "| Lowest mark:", lowest_mark)
print(f"4. Student having the highest total marks: Student {best_student}")
```

## 4. A grayscale image is represented by the following matrix. 
[[100,120,130], 
[150,180,200], 
[ 50, 80,100]] 
Perform: 
1. Normalize pixel values (divide by 255) 
2. Increase brightness by 30 
3. Clip values above 255 
4. Calculate average pixel intensity

```python
import numpy as np

# Grayscale image matrix
image = np.array([[100, 120, 130],
                  [150, 180, 200],
                  [ 50,  80, 100]])

# 1. Normalize pixel values (divide by 255)
normalized_image = image / 255.0

# 2. Increase brightness by 30
brightened_image = image + 30

# 3. Clip values above 255
clipped_image = np.clip(brightened_image, a_min=0, a_max=255)

# 4. Calculate average pixel intensity
average_intensity = np.mean(image)

print("1. Normalized pixel values:\n", np.round(normalized_image, 4))
print("\n2. Increased Brightness (by 30):\n", brightened_image)
print("\n3. Clipped values (max 255):\n", clipped_image)
print("\n4. Average pixel intensity:", round(average_intensity, 2))
```

```python
import numpy as np

# Monthly sales of 4 products over 6 months
sales = np.array([[120, 130, 140, 150, 160, 170],
                  [100, 110, 120, 130, 140, 150],
                  [150, 145, 160, 170, 180, 190],
                  [ 90, 100, 110, 120, 130, 140]])

# 1. Total sales of each product
total_sales_product = np.sum(sales, axis=1)

# 2. Average monthly sales of each product
avg_sales_product = np.mean(sales, axis=1)

# 3. Best-selling product
best_product = np.argmax(total_sales_product) + 1

# 4. Month with the highest overall sales
total_sales_month = np.sum(sales, axis=0)
best_month = np.argmax(total_sales_month) + 1

# 5. Standard deviation of sales for each product
std_dev_product = np.std(sales, axis=1)

print("1. Total sales of each product:", total_sales_product)
print("2. Average monthly sales of each product:", np.round(avg_sales_product, 2))
print(f"3. Best-selling product: Product {best_product}")
print(f"4. Month with the highest overall sales: Month {best_month}")
print("5. Standard deviation of sales for each product:", np.round(std_dev_product, 2))
```
