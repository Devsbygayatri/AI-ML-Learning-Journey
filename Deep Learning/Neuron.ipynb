```python
import torch
import torch.nn as nn  #nn- alias/building block of neural networks
```

```python
#input 
torch.manual_seed(42) #like random state->gives same output everytime(seed value fed)
inputs = torch.tensor([1.0, 2.0, 3.0]) #vector-1D array-> sent in float format
#tensor- multi-dimensional arrays used in pytorch
```

```python
# pytorch does NOT have 'step' activation fnx
neuron = nn.Linear(in_features = 3 , out_features = 1) #Linear Activation fnx->cals weighted sum
```

```python
output = neuron(inputs)
```

```python
output
```

```python
#verify ... formula(weighted sum)
# take dot product of input and weights + bias
neuron.weight @ inputs + neuron.bias  #'@'->dot product
#ans should be same with output
```

```python
neuron.weight # printing weights
```

```python
neuron.bias # printing bias
```
