```python
# Program
numbers = [12, 45, 7, 23, 56, 89, 2, 34, 67, 90]

largest = max(numbers)
smallest = min(numbers)
total_sum = sum(numbers)
average = total_sum / len(numbers)

print("List:", numbers)
print("Largest element:", largest)
print("Smallest element:", smallest)
print("Sum:", total_sum)
print("Average:", average)
```

# NESTED LIST

```python
# Program
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

second_row = matrix[1]
third_column = [row[2] for row in matrix]
total_sum = sum(sum(row) for row in matrix)

print("Second row:", second_row)
print("Third column:", third_column)
print("Sum of all elements:", total_sum)
```

```python
# Program
students = [
    ["Alice", 85, 90, 92], 
    ["Bob", 78, 81, 85], 
    ["Charlie", 95, 96, 99]
]

highest_avg = 0
topper = ""

for student in students:
    name = student[0]
    marks = student[1:]
    avg = sum(marks) / len(marks)
    print(f"{name}'s average: {avg:.2f}")
    
    if avg > highest_avg:
        highest_avg = avg
        topper = name

print("Student with highest average:", topper)
```

# Dictionary

```python
# Program
student = {
    "Name": "John Doe",
    "Roll Number": "CS101",
    "Department": "Computer Science",
    "CGPA": 8.5
}

print("Original Details:")
for key, value in student.items():
    print(f"{key}: {value}")

# Update CGPA
student["CGPA"] = 9.1

print("\nUpdated Details:")
print("CGPA:", student["CGPA"])
```

```python
# Program
marks = {"Math": 95, "Science": 88, "English": 92}

print("Keys:", list(marks.keys()))
print("Values:", list(marks.values()))

# Subject with highest marks
highest_subject = max(marks, key=marks.get)
print("Subject with highest marks:", highest_subject)
```

```python
# Program
student_data = {"Name": "Grace", "Marks": [85, 90, 78, 92, 88]}

marks_list = student_data["Marks"]
average_marks = sum(marks_list) / len(marks_list)
highest_mark = max(marks_list)
lowest_mark = min(marks_list)

print("Student Name:", student_data["Name"])
print("Average Mark:", average_marks)
print("Highest Mark:", highest_mark)
print("Lowest Mark:", lowest_mark)
```

# Dictionary in List

```python
# Program
students = [
    {"Name": "Alice", "Marks": 88},
    {"Name": "Bob", "Marks": 95},
    {"Name": "Charlie", "Marks": 82}
]

print("Student Names:")
for s in students:
    print(s["Name"])

topper = max(students, key=lambda x: x["Marks"])
print(f"The topper is {topper['Name']} with {topper['Marks']} marks.")
```

```python
# Program
books = [
    {"Title": "Python Crash Course", "Price": 30},
    {"Title": "Clean Code", "Price": 45},
    {"Title": "Fluent Python", "Price": 50}
]

expensive_book = max(books, key=lambda x: x["Price"])
total_cost = sum(b["Price"] for b in books)

print(f"Most expensive book: {expensive_book['Title']} (Rupees {expensive_book['Price']})")
print(f"Total cost of all books: Rupees {total_cost}")
```

# SET

```python
# Program
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference (set1 - set2):", set1 - set2)
print("Symmetric Difference:", set1 ^ set2)
```
